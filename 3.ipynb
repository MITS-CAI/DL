```python
#EXP-3.1: Implementing Our First Neuron
inputs = [1, 2, 3]
weights = [0.2, 0.8, -0.5]
bias = 2
# Multiply inputs by weights, add the results and bias
output = sum(i * w for i, w in zip(inputs, weights)) + bias
print(output)
#EXP-3.2: Single-Layered Neural Network with 2 Input Values
import numpy as np
inputs = np.array([2, 3])
w1 = [1, 1]
w2 = [-1, 1]
w3 = [2, -1]
# Hidden layer
x2 = np.dot(inputs, w1)
x3 = np.dot(inputs, w2)
hidden = np.array([x2, x3])
# Output layer
x4 = np.dot(hidden, w3)
print("Input Layer:", inputs)
print("Hidden Layer:", hidden)
print("Output Layer:", x4)
#EXP-3.3: Multi-Layered Neural Network with 2 Input Values
import numpy as np
inputs = np.array([2, 3])
# First hidden layer
x2 = np.dot(inputs, [1, 1])
x3 = np.dot(inputs, [-1, 1])
hidden1 = np.array([x2, x3])
# Second hidden layer
x4 = np.dot(hidden1, [-2, 1])
x5 = np.dot(hidden1, [1, 0])
hidden2 = np.array([x4, x5])
# Output layer
x6 = np.dot(hidden2, [1, -1])
print("Input Layer:", inputs)
print("Hidden Layer 1:", hidden1)
print("Hidden Layer 2:", hidden2)
print("Output Layer:", x6)
```
