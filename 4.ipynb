```python

def relu(x):
  return max(0,x)

```

```python
import numpy as np
def sigmoid(x):
  return 1/(1+np.exp(-x))
```

```python
import numpy as np
input=np.array([2,3])
x2=np.array([1,1])
x3=np.array([-1,1])
x4=np.array([-2,1])
x5=np.array([1,0])
x6=np.array([1,-1])
x2=np.dot(input,x2)
x3=np.dot(input,x3)
x2=relu(x2)
x3=relu(x3)
h1=np.array([x2,x3])
print(h1)
x5=np.dot(h1,x5)
x4=np.dot(h1,x4)
x5=relu(x5)
x4=relu(x4)
h2=np.array([x4,x5])
print(h2)
#output layer
x6=np.dot(h2,x6)
x6=sigmoid(x6)
output=np.array([x6])

print("output= ",output)
```
