```python
import cv2
from google.colab.patches import cv2_imshow

# Load face detector
face_cascade = cv2.CascadeClassifier("/content/haarcascade_frontalface_default (1).xml")
# Built-in option (no XML upload needed):
# face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")

# Read color image (for final display) and grayscale image (using 0, for detection)
img = cv2.imread("/content/phlimages.jpg")
gray = cv2.imread("/content/phlimages.jpg", 0)

# Detect faces and draw green rectangles on the color image
faces = face_cascade.detectMultiScale(gray, 1.05, 5)
for x, y, w, h in faces:
    cv2.rectangle(img, (x, y), (x + w, y + h), (0, 255, 0), 3)

# Show result
cv2_imshow(img)
```
