```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

# 1. Load dataset and split into features (x) & target (y)
df = pd.read_csv("/content/housepricedata.csv")
x = df.iloc[:, 0:10]
y = df.iloc[:, 10]

# 2. Scale features (0 to 1) and split into train/test sets
scaler = MinMaxScaler()
x_scaled = scaler.fit_transform(x)
x_train, x_test, y_train, y_test = train_test_split(x_scaled, y, test_size=0.2, random_state=2)

# 3. Build ANN model
model = Sequential([
    Input(shape=(10,)),
    Dense(10),
    Dense(20, activation="relu"),
    Dense(20, activation="relu"),
    Dense(1, activation="sigmoid")
])

# 4. Compile & train
model.compile(loss="binary_crossentropy", optimizer="sgd", metrics=["accuracy"])
model.fit(x_train, y_train, batch_size=10, epochs=110, validation_data=(x_test, y_test), verbose=0)

# 5. Evaluate accuracy & predict on a new house
loss, acc = model.evaluate(x_test, y_test, verbose=0)
print("Test Accuracy:", acc)

sample = pd.DataFrame([[23, 3, 4, 56565, 676, 7, 2, 3, 4, 5]], columns=x.columns)
result = model.predict(scaler.transform(sample), verbose=0)
print("Prediction:", result)
```
