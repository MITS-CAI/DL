```python
import cv2
import matplotlib.pyplot as plt
# Read image in grayscale
img = cv2.imread("/content/apple.webp", 0)
# Apply different thresholding methods
_, t1 = cv2.threshold(img, 125, 255, cv2.THRESH_BINARY)
_, t2 = cv2.threshold(img, 125, 255, cv2.THRESH_BINARY_INV)
_, t3 = cv2.threshold(img, 125, 255, cv2.THRESH_TOZERO)
_, t4 = cv2.threshold(img, 125, 255, cv2.THRESH_TOZERO_INV)
_, t5 = cv2.threshold(img, 125, 255, cv2.THRESH_TRUNC)
# Display results
plt.figure(figsize=(10, 10))
plt.subplot(231)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.subplot(232)
plt.imshow(t1, cmap="gray")
plt.title("THRESH_BINARY")
plt.subplot(233)
plt.imshow(t2, cmap="gray")
plt.title("THRESH_BINARY_INV")
plt.subplot(234)
plt.imshow(t3, cmap="gray")
plt.title("THRESH_TOZERO")
plt.subplot(235)
plt.imshow(t4, cmap="gray")
plt.title("THRESH_TOZERO_INV")
plt.subplot(236)
plt.imshow(t5, cmap="gray")
AI / ML Lab Experiments Page 3
plt.title("THRESH_TRUNC")
plt.tight_layout()
plt.show()
```
